###**Check Outliers:**

```python
#IQR-based Outlier Counts:
numeric_cols = df.select_dtypes(include='number').columns

for col in numeric_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    count = ((df[col] < lower) | (df[col] > upper)).sum()
    print(f"{col}: {count} outliers")
```

```python
#Check BMI Outlier:
Q1 = df['BMI'].quantile(0.25)
Q3 = df['BMI'].quantile(0.75)
IQR =Q3-Q1
Lowerbound = Q1-1.5 * IQR
Upperbound = Q3+1.5 * IQR
print("Lowerbound",Lowerbound)
print("Upperbound",Upperbound)
```

**Analysis:**
- BMI values below 12.45 are considered lower outliers.
- BMI values above 43.33 are considered upper outliers.
- Values between these boundaries are not outliers according to the IQR method.

```python
Outliers = df[(df['BMI']<Lowerbound) |(df['BMI']>Upperbound)]
print("Total BMI Outliers:",len(Outliers))
```

```python
outlier_percentage = len(Outliers) / len(df)* 100
print("Outlier Percentage:", outlier_percentage)
```

**BMI Outlier Analysis:**
- I used the Interquartile Range (IQR) method to identify outliers in the BMI column.
- The lower and upper bounds were 12.45 and 43.33, respectively. I identified 8,905 outliers, representing approximately 2.91% of the dataset.
- Since extreme BMI values may represent genuine health observations, I decided to retain them rather than remove them without further investigation.
